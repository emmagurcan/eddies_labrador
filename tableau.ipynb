```python
def get_reference_profile(cruise, section_num, ds, dss, lon, lat, CT, SA, distance, depth, sub_section=0):
    if cruise == "MSM74" and (section_num == 2 or section_num == 5):
        lon_woa = ds['lon']
        lat_woa = ds['lat']
        t_mn = np.squeeze(dss['s_mn'])
        s_mn = np.squeeze(ds['t_mn'])
        depth_ref = np.squeeze(ds['depth'])
        sref, lon_sel, lat_sel = get_woa_ref(lon, lat, lon_woa, lat_woa, t_mn)
        tref, lon_woa_sels, lat_woa_sels = get_woa_ref(lon, lat, lon_woa, lat_woa, s_mn)
        inf_mask = np.where(tref == 9.9692100e+36)
        tref[inf_mask] = np.nan

        inf_mask = np.where(sref == 9.9692100e+36)
        sref[inf_mask] = np.nan
    
    if cruise == "MSM40" and (section_num == 3 or section_num == 4 or (section_num == 7 and sub_section==3 )):
        lon_woa = ds['lon']
        lat_woa = ds['lat']
        t_mn = np.squeeze(dss['s_mn'])
        s_mn = np.squeeze(ds['t_mn'])
        depth_ref = np.squeeze(ds['depth'])
        sref, lon_sel, lat_sel = get_woa_ref(lon, lat, lon_woa, lat_woa, t_mn)
        tref, lon_woa_sels, lat_woa_sels = get_woa_ref(lon, lat, lon_woa, lat_woa, s_mn)
        inf_mask = np.where(tref == 9.9692100e+36)
        tref[inf_mask] = np.nan

        inf_mask = np.where(sref == 9.9692100e+36)
        sref[inf_mask] = np.nan
    
    if cruise == "MSM40" and section_num == 7 and sub_section==2:
        woa_salinity_path1 = "/Users/emmagurcan/Documents/France/ENS/M1/stageM1/data/woa/august/2015-2022_mean/woa23_decav_s08_01.nc"
        woa_path1 = "/Users/emmagurcan/Documents/France/ENS/M1/stageM1/data/woa/august/2015-2022_mean/woa23_B5C2_t08_01.nc"
        ds1 = nc.Dataset(woa_path1)
        dss1 = nc.Dataset(woa_salinity_path1)
        lon_woa = ds1['lon']
        lat_woa = ds1['lat']
        t_mn = np.squeeze(dss1['s_mn'])
        s_mn = np.squeeze(ds1['t_mn'])
        depth_ref = np.squeeze(ds['depth'])
        sref, lon_sel, lat_sel = get_woa_ref(lon, lat, lon_woa, lat_woa, t_mn)
        tref, lon_woa_sels, lat_woa_sels = get_woa_ref(lon, lat, lon_woa, lat_woa, s_mn)
        inf_mask = np.where(tref == 9.9692100e+36)
        tref[inf_mask] = np.nan

        inf_mask = np.where(sref == 9.9692100e+36)
        sref[inf_mask] = np.nan

    if cruise == "MSM74" and (section_num == 3 or section_num == 4):
        tref, lon_sel, lat_sel = create_ref_profile(CT, distance, lon, lat, 0, 450)
        sref, lon_sel, lat_sel = create_ref_profile(SA, distance, lon, lat, 0, 450)
        depth_ref = depth
    
    if cruise == "MSM74" and section_num == 5:
        tref, lon_sel, lat_sel = create_ref_profile(CT, distance, lon, lat, 200, 450)
        sref, lon_sel, lat_sel = create_ref_profile(SA, distance, lon, lat, 200, 450)
        depth_ref = depth
    
    if cruise == "MSM74" and section_num == 7:
        tref, lon_sel, lat_sel = create_ref_profile(CT, distance, lon, lat, 100, 450)
        sref, lon_sel, lat_sel = create_ref_profile(SA, distance, lon, lat, 100, 450)
        depth_ref = depth

    if cruise == "MSM40" and (section_num == 1 or section_num == 5 or (section_num == 7 and sub_section ==1)):
        tref, lon_sel, lat_sel = create_ref_profile(CT, distance, lon, lat, 0, 450)
        sref, lon_sel, lat_sel = create_ref_profile(SA, distance, lon, lat, 0, 450)
        depth_ref = depth
    
    # if cruise == "MSM74" and section_num ==7:
    #     print("SECTION 7 MSM74 REFERENCE PROFILE")
    #     woa_salinity_path1 = "/Users/emmagurcan/Documents/France/ENS/M1/stageM1/data/woa/august/2015-2022_mean/woa23_decav_s08_01.nc"
    #     woa_path1 = "/Users/emmagurcan/Documents/France/ENS/M1/stageM1/data/woa/august/2015-2022_mean/woa23_B5C2_t08_01.nc"
    #     ds1 = nc.Dataset(woa_path1)
    #     dss1 = nc.Dataset(woa_salinity_path1)
    #     lon_woa = ds1['lon']
    #     lat_woa = ds1['lat']
    #     t_mn = np.squeeze(dss1['s_mn'])
    #     s_mn = np.squeeze(ds1['t_mn'])
    #     depth_ref = np.squeeze(ds['depth'])
    #     sref, lon_sel, lat_sel = get_woa_ref(lon, lat, lon_woa, lat_woa, t_mn)
    #     tref, lon_woa_sels, lat_woa_sels = get_woa_ref(lon, lat, lon_woa, lat_woa, s_mn)
    #     inf_mask = np.where(tref == 9.9692100e+36)
    #     tref[inf_mask] = np.nan

    #     inf_mask = np.where(sref == 9.9692100e+36)
    #     sref[inf_mask] = np.nan


    return tref, sref, lon_sel, lat_sel, depth_ref 

def detect_eddy(u_adcp, v_adcp, depth_adcp, lon, lat, distance):
    depth_dim, dist_dim = u_adcp.shape

    # Compute the magnitude of horizontal velocity
    speed = np.sqrt(u_adcp**2 + v_adcp**2)

    # Compute direction of velocity vectors
    angle = np.arctan2(v_adcp, u_adcp)  # angle in radians

    # Derivative of angle along distance (angular shear)
    dtheta_dx = np.gradient(angle, axis=1) / np.gradient(distance)

    # Identify candidate eddy centers as where rotation changes sign and speed is low
    eddy_centers = []
    for z in range(depth_dim):
        # Zero-crossings in angular derivative may suggest a rotation center
        sign_change = np.diff(np.sign(dtheta_dx[z]))
        zero_cross_indices = np.where(sign_change != 0)[0]
        for idx in zero_cross_indices:
            if speed[z, idx] < np.percentile(speed[z], 20):  # threshold for low velocity
                eddy_centers.append((z, idx))

    if not eddy_centers:
        print("No eddy detected.")
        return None

    # Choose the best eddy center (lowest velocity)
    zc, ic = min(eddy_centers, key=lambda x: speed[x])

    eddy_center_lon = lon[ic]
    eddy_center_lat = lat[ic]
    center_depth = depth_adcp[zc]

    # Radius estimation: distance from center to max velocity
    profile_speed = speed[zc]
    from scipy.signal import find_peaks
    peaks, _ = find_peaks(profile_speed)
    if len(peaks) == 0:
        print("No clear velocity peak for radius estimation.")
        return None

    # Choose the nearest peak from the center
    peak_idx = peaks[np.argmin(np.abs(peaks - ic))]
    eddy_radius = np.abs(distance[peak_idx] - distance[ic])
    max_velocity = profile_speed[peak_idx]

    # Determine depth extent by checking coherence of rotation
    coherent_depths = []
    for z in range(depth_dim):
        if ic > 0 and ic < dist_dim - 1:
            left = dtheta_dx[z, ic - 1]
            right = dtheta_dx[z, ic + 1]
            if np.sign(left) != np.sign(right):
                coherent_depths.append(depth_adcp[z])

    if not coherent_depths:
        eddy_depth_extent = (center_depth, center_depth)
    else:
        eddy_depth_extent = (min(coherent_depths), max(coherent_depths))

    results = {
        'lon': eddy_center_lon,
        'lat': eddy_center_lat,
        'center_depth': center_depth,
        'eddy_radius': eddy_radius,
        'max_velocity': max_velocity,
        'depth_extent': eddy_depth_extent
    }

    print("Eddy detection results:")
    for k, v in results.items():
        print(f"{k}: {v}")

    return results
```
